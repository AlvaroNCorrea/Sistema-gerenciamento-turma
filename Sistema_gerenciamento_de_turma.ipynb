```python
turmaad6 = {
     # chave é unitario, valor pode não ser unitario
     "professores" : ["Felipe", "Luiz", "Priscilla"],
     "alunos" : ["Andre", "Paulo", "Aline", "Paola"]
}

print(turmaad6['professores'][1])
print(turmaad6['alunos'][2])
# vamos criar funções
# ver os dados
#adicionar novo professor
# adicionar novo aluno
#remover aluno
# remover professor
def ver_dados():
  print("---- Professores cadastrados ----")
  for i in turmaad6['professores']:
    print(i)

  print('---- alunos cadastrados ----')
  for i in turmaad6['alunos']:
    print(i)

def add_prof():
  novo_professor = input('Qual o nome do professor: ')
  # chama a lista .append(dado)
  turmaad6['professores'].append(novo_professor)
  print(f'o professor {novo_professor} foi adcionado')
  ver_dados()

add_prof()
```
